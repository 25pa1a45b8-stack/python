```python
# Write mode
file = open("sample.txt", "w")
file.write("Hello Python\n")
file.close()

# Append mode
file = open("sample.txt", "a")
file.write("Welcome to file handling")
file.close()

# Read mode
file = open("sample.txt", "r")
print(file.read())
file.close()
```

```python
file = open("sample.txt", "r")
lines = file.readlines()
file.close()

for line in lines[::-1]:
    print(line.strip())
```

```python
file = open("sample.txt", "r")
text = file.read()
file.close()

characters = len(text)
words = len(text.split())
lines = len(text.splitlines())

print("Characters:", characters)
print("Words:", words)
print("Lines:", lines)
```

```python
file = open("input.txt", "r")
text = file.read()
file.close()

words = text.lower().split()
words.sort()

file = open("output.txt", "w")
for word in words:
    file.write(word + "\n")
file.close()

print("Words sorted and saved")
```

```python
a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        result[i][j] = a[i][j] + b[i][j]

for row in result:
    print(row)
```

```python
a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] = result[i][j] + a[i][k] * b[k][j]

for row in result:
    print(row)
```
