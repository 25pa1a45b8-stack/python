```python
def palindrome(number):
    reverse = 0
    temp = number

    while temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp = temp // 10

    if number == reverse:
        return True
    else:
        return False

n = int(input("Enter a number: "))

if palindrome(n):
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
def reverse_string(text):
    return text[::-1]

word = input("Enter a string: ")
print("Reverse string:", reverse_string(word))
```

```python
def student(name, course="Python"):
    print("Name:", name)
    print("Course:", course)

student("Ravi")
student("Anu", "Java")
```

```python
def calculate(a, b):
    sum = a + b
    difference = a - b
    product = a * b
    return sum, difference, product

x, y, z = calculate(10, 5)
print("Sum:", x)
print("Difference:", y)
print("Product:", z)
```

```python
numbers = [1, 2, 2, 3, 3, 3, 4]
duplicates = 0
checked = []

for num in numbers:
    if numbers.count(num) > 1 and num not in checked:
        duplicates = duplicates + 1
        checked.append(num)

print("Number of duplicate elements:", duplicates)
```

```python
marks = {"Maths": 80, "Science": 90, "English": 70}
total = 0

for value in marks.values():
    total = total + value

print("Sum of dictionary values:", total)
```

```python
numbers = [1, 2, 2, 3, 1, 2]
frequency = { }

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print(frequency)
```
